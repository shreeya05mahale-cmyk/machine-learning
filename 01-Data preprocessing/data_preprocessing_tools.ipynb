# Data Preprocessing Tools

## Importing the libraries


```
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Importing the dataset

```
dataset = pd.read_csv('Data.csv')

#Include all rows, include all columns except the last one
X = dataset.iloc[:, :-1].values #Features

#Include all rows, include only the last column
y = dataset.iloc[:, -1].values #Dependent variable
```

```
print(X)
```

```
print(y)
```

## Taking care of missing data

```
from sklearn.impute import SimpleImputer

#Create an object of SimpleImputer class
imputer = SimpleImputer(missing_values=np.nan, strategy="mean")

#Include all rows, include column 1 and 2
imputer.fit(X[:, 1:3])

#transform() returns a dataframe
X[:, 1:3] = imputer.transform(X[:, 1:3])
```

```
print(X)
```

## Encoding categorical data

### Encoding the Independent Variable

```
# Country column has France Germany Spain, a machine learning algorithm cannot
# work directly with words

#For applying a particular transformation to particular columns
from sklearn.compose import ColumnTransformer
#Transforms France Germany Spain into separate columns (with yes or no questions)
from sklearn.preprocessing import OneHotEncoder

ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [0])], remainder='passthrough')
#'encoder' is the name given to the transformation
#Apply one-hot encoding => France becomes 1 0 0 and so on
#Apply one-hot encoding only to column[0]
#For the columns not specified, leave them as they are

#fit() learns the rules, transforms() learns the transformations and applies it
X = np.array(ct.fit_transform(X)) #ColumnTransformer returns a special matrix representation
```

```
print(X)
```

### Encoding the Dependent Variable

```
#Purchased column has Yes No classes
#LabelEncoder assigns numbers to each class
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y = le.fit_transform(y)
```

```
print(y)
```

## Splitting the dataset into the Training set and Test set

```
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 1)
#With random_state we get the same split each
#The values for random_state is like the starting point of a shuffle, it does not indicate
#the amount of randomness
```

```
print(X_train)
```

```
print(X_test)
```

```
print(y_train)
```

```
print(y_test)
```

## Feature Scaling

```
#StandardScaler performs standardization
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()

#fit() calculates mean and standard deviation
#transform() applies the changes to the dataset
X_train[:, 3:] = sc.fit_transform(X_train[:, 3:])

#We use only transform() for the test set because we want the exact same scale as the training set
#Predictions are made relative to our training set
X_test[:, 3:] = sc.transform(X_test[:, 3:])
```
